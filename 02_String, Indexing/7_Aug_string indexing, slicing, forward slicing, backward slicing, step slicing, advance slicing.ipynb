# Step Slicing

```python
str1 = "Welcome"
str1
```

```python
str1[0:7:2]
```

here above
* 0123456
* Welcome
* -7-6-5-4-3-2-1
* for str1[0:7:3] ( for every 2 steps, from 0 to 7 )
* 0-'W', then again skip two steps(We) next is... 'l', then again skip two steps(lc) next is...   'o', then (om) next is...   'e'
* final output is 'Wloe'

```python
str1[-5:-1]
```

here
* left index -5 = l but right index changes because (n-1) for right index
* n = right index value then (n-1)
* (-1-1) = -2
* -2 = m, so for [-5:-1]
* output = 'lcom'

```python
str1[::2] # every second index slice from starting to ending
```

```python
str1[::-1]
```
