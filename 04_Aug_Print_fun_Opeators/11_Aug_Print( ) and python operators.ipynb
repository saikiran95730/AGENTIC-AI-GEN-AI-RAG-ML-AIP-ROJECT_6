## print() with format method

```python
num1 =  30
num2 = 40
num3 = 50
num4 = 60

add = num1 + num2 + num3 + num4

print('The addition of {} and {} and {} and {} is {}'.format(num1,num2,num3,num4,add))
```

## print() with f string method

```python
num1 =  30
num2 = 40
num3 = 50
num4 = 60

add = num1 + num2 + num3 + num4

print(f'The addition of {num1} and {num2} and {num3} and {num4} is {add}')
```
