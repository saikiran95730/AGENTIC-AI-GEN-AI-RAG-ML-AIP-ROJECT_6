### 1 . input( ) with number

```python
x1 = int(input('firstNumber'))
y1 = int(input('secondNumber'))
z1 = x1 + y1
print(z1)
```

### 2 . input( ) with string & string indexing2 

```python
strin = input('enter a string')[0]
print(strin)
```

```python
strin = input('enter a string')[5]
print(strin)
```

```python
strin = input('enter a string')[3:]
print(strin)
```

```python
strin = input('enter a string')[5:-1]
print(strin)
```

* in prompt engineering we will be using this input( ) a lot
