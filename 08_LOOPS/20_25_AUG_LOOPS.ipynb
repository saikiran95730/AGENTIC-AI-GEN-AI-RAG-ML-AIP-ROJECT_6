## WHILE LOOP

```python
i = 1    # initializing

while i <= 5:  # condition
    print('Data science', i)
    i = i + 1  # increment
```

```python
i = 5    # initializing

while i >= 1:  # condition
    print('Data science', i )
    i = i - 1  # decrement
```

```python
i = 5    # initializing

while i >= 1:  # condition
    print('Data science', i, end = ' - ') # when we mention end then new Line will not create
    
    j = 1
    while j<= 4:
        print('Full stack', j, end = ' - ')
        j  = j+ 1
             
    i = i - 1  # decrement
     
# the output which we got is very lengty but how to make them one line Lets refer to below code
```

## FOR Loops

```python
for i in range(5):
    print(i)
```

```python
for i in range(1,10,3):
    print(i)
```

```python
# print value which is divisible by 5

for i in range(1,51):
    if i%5 == 0:
        print(i)
```

  #### In for loop we have some powerful keywords

- break
- continue
- pass

### break

```python
for i in range(1,51):
    
    if i == 5:
        break  # break the loop
    print(i)  # print(i) is inside the loop
```

```python
for i in range(1,51):
    
    if i == 5:
        break  # break the loop
print(i)  # print(i) is outside the loop and it prints only last number where it stops
```

### continue

```python
for i in range(1,10):
    
    if i == 5:
        continue # skips the specific given condition
    print(i) 
```

```python
for i in range(1,10):
    
    if i == 5:
        continue  # skips the specific given condition
print(i)  # print(i) is outside the loop and it prints only last number where it stops
```

### pass

```python
for i in range(1,10):
```

```python
for i in range(1,10):
    pass
```

```python
for i in range(1,51):
    if i % 3 == 0 or i % 5 == 0:
        continue
    print(i)
print('end')
```

### # PRINTING PATTERN in PYTHON

```python
for i in range(5):
    for j in range(i+1):
        print('#', end = ' ')
    print()
```

```python
nums = [12, 18, 15, 20]
for i in nums:
    if i % 5 == 0:
        print(i)
    
```

```python
nums = [12, 18, 15, 20]
for i in nums:
    if i % 5 == 0:
        break
        print(i)
```

```python
nums = [12, 18, 15, 20]
for i in nums:
    if i % 5 == 0:
        print(i)
        break
```

```python
nums = [12, 18, 17, 21, 26]
for i in nums:
    if i % 5 == 0:
        print(i)
        break
    else:
        print("number not found")
```

### for..else

```python
num = 13
for i in range(2, num):
    if num % i == 0:
        print('not a prime num')
        break
else:
    print('prime num')
```
