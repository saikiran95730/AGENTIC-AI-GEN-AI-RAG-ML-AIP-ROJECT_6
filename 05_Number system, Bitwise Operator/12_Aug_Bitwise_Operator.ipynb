# Bitwise_Operator

### 1.   ~ (Compliment)

```python
~12 #  ~n = -(n+1)
```

```python
~150  #  -(150+1)---   -151
```

```python
~(-121)  # -(-121+1)---   120
```

### 2.   & (AND)

```python
12 & 13
```

```python
20 & 25
```

### 3. | (OR)

```python
12 | 13 # same as AND but (1 or 0 -- 1), (0 or 1 -- 1) 
```

### 4. ^ (XOR)

##### 0,0 -- 0 // 1,1 -- 0   // same same - 0 
##### 1,0 -- 1 // 0,1 -- 1   // Different - 1 

```python
12 ^ 13
```

### 5. << (Left shift operator)

#### we will gain the bit if user mention 2 then 2 bits, 3 -- 3bits, 4- 4bits,.....

```python
12 << 2
```

```python
12 << 1
```

### 6. >> (Right shift operator)

#### we will loose the bit (remove the bits)

```python
12 >> 1
```

```python
12 >> 2 
```
