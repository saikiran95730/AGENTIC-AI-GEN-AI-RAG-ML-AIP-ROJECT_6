# Swap 2 Variable

## 1 way


```python
a = 5
b = 7
```

```python
print(a)
print(b)
```

* This is one way of swapping

```python
temp = a  # temp = 5
a = b     # a = 7
b = temp  # b = 5
```

```python
print(a)
print(b)
```

##  2 way

```python
a1 = 5
b1 = 7
```

```python
print(a1)
print(b1)
```

* This is another way of swapping

```python
a1 = a1 + b1  # a1 = 5  + 7 = 11 ---- a1 = 11
b1 = a1 - b1  # b1 = 11 - 7 = 5  ---- b1 = 5
a1 = a1 - b1  # a1 = 11 - 5 = 7  ---- a1 = 7
```

```python
print(a1)
print(b1)
```

##  3 way

```python
a2 = 5
b2 = 7
```

```python
print(a2)
print(b2)
```
