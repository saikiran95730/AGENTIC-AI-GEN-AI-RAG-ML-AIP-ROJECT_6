## 6. Superset

```python
a2 = {1, 2, 3, 4, 5, 6, 7, 8, 9}
b2 = {3, 4, 5, 6, 7, 8}
c2 = {10, 20, 30, 40}
```

```python
a2.issuperset(b2) # whatever values present in a2 are present in b2 too. a2 is bigger SET than b2. So a2 is superset of b2
```

## 7. Subset

```python
 b2.issubset(a2) # All values present in b2 are also present in a2 too and b2 is smaller than a2. so b2 is subset to a2
```

## 8. Disjoint

```python
a3 = {1, 2, 3, 4, 5, 6, 7, 8}
b3 = {40, 50, 50, 60}
c3 = {10, 20, 30}
```

```python
a3.isdisjoint(b3) # whatever values present in a2 are not at all present in b2. So a2 is disjoint of b2
```

```python
b3.isdisjoint(c3)
```

```python
a3.issubset(b3)
```

```python
b3.issuperset(c3)
```
